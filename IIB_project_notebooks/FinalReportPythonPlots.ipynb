```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import levy_stable

# 1) α‐stable density varying over β at fixed α=0.5, σ=1, μ=0
x1 = np.linspace(-4, 4, 2001)
alpha1 = 0.5
scale1 = 1.0
loc1 = 0.0
betas = [0.0, 0.25, 0.5, 0.75, 1.0]
colors_beta = ['black','blue','green','red','magenta']

plt.figure(figsize=(6,4))
for β, c in zip(betas, colors_beta):
    pdf = levy_stable.pdf(x1, alpha1, β, loc=loc1, scale=scale1)
    plt.plot(x1, pdf, color=c, label=f'β = {β:.2f}', lw=1.5)
plt.ylim(0,0.7)
plt.xlim(-4,4)
plt.xlabel('x')
plt.ylabel('p(x)')
plt.legend(title=r'$\beta$')
# annotate α,σ,μ
plt.text(2.0, 0.65, r'$\alpha=0.5,\ \sigma=1,\ \mu=0$', 
         ha='right', va='top', fontsize= 9)
plt.title(r'$\alpha$‐stable: varying skew $\beta$')
plt.tight_layout()
plt.show()

# 2) α‐stable density varying over α at fixed β=0, σ=1, μ=0
x2 = np.linspace(-6, 6, 3001)
alphas = [2.0, 1.0, 0.5]   # Gaussian, Cauchy, Lévy
colors_alpha = ['blue','green','red']

plt.figure(figsize=(6,4))
for α, c in zip(alphas, colors_alpha):
    pdf = levy_stable.pdf(x2, α, 0.0, loc=0.0, scale=1.0)
    plt.plot(x2, pdf, color=c, label=fr'$\alpha = {α:g}$', lw=1.5)
plt.ylim(0,0.8)
plt.xlim(-6,6)
plt.xlabel('X')
plt.ylabel('p(X)')
plt.legend(loc='upper right')
# annotate arrows
plt.annotate('Gaussian', xy=(-4.5, 0.1), xytext=(-5.5,0.25),
             arrowprops=dict(arrowstyle='->'))
plt.annotate('Cauchy',   xy=(-2.0, 0.25), xytext=(-3.0,0.4),
             arrowprops=dict(arrowstyle='->'))
plt.annotate('Lévy',     xy=(0.5,  0.65), xytext=(1.0,0.75),
             arrowprops=dict(arrowstyle='->'))
plt.title(r'$\alpha$‐stable: varying tail‐index $\alpha$')
plt.tight_layout()
plt.show()
```
